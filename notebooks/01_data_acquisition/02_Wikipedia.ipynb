### Getting Data from a Website

In this case we are going to extract a atable from [Wikipedia](https://en.wikipedia.org/wiki/Climate_of_Malta)
This is done by obtaining the HTML and extracting the desired element.

python -m pip install bs4 requests

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

URL = 'https://en.wikipedia.org/wiki/Climate_of_Malta'
```

```python
response = requests.get(
    url=URL,
    headers={"User-Agent" : "MCAST-ClassDemo"},
    timeout=30
)
response.raise_for_status()

response.content
```

```python
soup = BeautifulSoup(response.content,'html.parser')
```

```python
table = soup.select_one("table.wikitable")
table
```
