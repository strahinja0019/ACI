## Boxplot

from a quick look we notice that year is a sequential number (no need to anayse)

Passangers variable appears to be large in numbers, hencfe we can create a visual to help us
understand its distrubution and general statictics

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(800,600,'px'))
plt.boxplot(df['passengers'], orientation='horizontal')
plt.title("Boxplot for Passengers")
plt.xlabel("passangers")
plt.yticks([])
plt.show()
```

## Histograms and Scatterplots

These can be generated automatically for all column in dataset (also against each other) using seaborn pairplot

```python
import seaborn as sns

sns.pairplot(df)
plt.show()
```

## Line Chart

A line chart is more suitable for a time-series chart. In this case we can see how passangers changed over the years.

```python
plt.figure(figsize=(8,6))
plt.plot(df['year'],df['passengers'])
plt.title("Passangers over years")
plt.xlabel("Year")
plt.ylabel("Passangers in Million")
plt.xticks(df['year'],rotation = 45)
plt.grid(axis='both')
```
