### Machine learning in python

```python
import pandas as pd

houses = pd.DataFrame({
    "floor_area_m2" : [70,90,110,140,170],
    "bedrooms": [1,2,2,3,4],
    "price_eur":[180000,230000,270000,340000,410000]
})

display(houses)
```

```python
houses.plot.scatter(
    x="floor_area_m2",
    y="price_eur"
)

import matplotlib.pyplot as plt

plt.xlabel("Floor Area (m2)")
plt.ylabel("House Price (euro)")
plt.title("Example House prices")

```
